Signal to Noise Ratio

```python
# Set FSL environment
fsl_dir = os.environ['FSLDIR']
bet_cmd = os.path.join(fsl_dir, 'bin', 'bet')
fslmaths_cmd = os.path.join(fsl_dir, 'bin', 'fslmaths')

# Input file
input_file = 'anat.nii.gz'  
output_file = 'brain.nii.gz'
noise_file = 'noise.nii.gz'

# Step 1: Skull-strip the image using BET (Brain Extraction Tool)
subprocess.run([bet_cmd, input_file, output_file, '-f', '0.3', '-g', '0'])

# Step 2: Calculate the mean signal inside the brain
brain_img = nib.load(output_file)
brain_data = brain_img.get_fdata()
mean_signal = np.mean(brain_data[brain_data > 0])  # Only consider non-zero voxels

# Step 3: Estimate the noise by subtracting the brain-extracted image from the original
subprocess.run([fslmaths_cmd, input_file, '-sub', output_file, noise_file])

# Step 4: Calculate the standard deviation of the noise
noise_img = nib.load(noise_file)
noise_data = noise_img.get_fdata()
std_noise = np.std(noise_data[noise_data > 0])  # Only consider non-zero voxels

# Step 5: Calculate SNR
snr = mean_signal / std_noise
print(f'Signal-to-Noise Ratio (SNR): {snr}')
```

```python
import nibabel as nib

# Set your file paths
dti_file = 'dti.nii.gz'
bvals_file = 'dti.bval'
bvecs_file = 'dti.bvec'

# Load the DTI file to check its dimensions
dti_img = nib.load(dti_file)
dti_shape = dti_img.shape
num_volumes = dti_shape[-1]

# Function to check the shape of bvals file
def check_bvals_shape(bvals_file, num_volumes):
    with open(bvals_file, 'r') as f:
        bvals = f.read().split()
    print(f"Number of b-values: {len(bvals)}")
    if len(bvals) != num_volumes:
        print(f"Error: The number of b-values ({len(bvals)}) does not match the number of volumes in the DTI file ({num_volumes}).")
    else:
        print("bvals file shape is correct.")

# Function to check the shape of bvecs file
def check_bvecs_shape(bvecs_file, num_volumes):
    with open(bvecs_file, 'r') as f:
        bvecs = [line.split() for line in f.readlines()]
    print(f"Number of rows in bvecs: {len(bvecs)}")
    print(f"Number of columns in each bvecs row: {len(bvecs[0])}")
    if len(bvecs) != 3 or len(bvecs[0]) != num_volumes:
        print(f"Error: The bvecs file shape ({len(bvecs)}x{len(bvecs[0])}) does not match the expected shape (3x{num_volumes}).")
    else:
        print("bvecs file shape is correct.")

# Check DTI file shape
print(f"DTI file shape: {dti_shape}")
print(f"Number of volumes in DTI file: {num_volumes}")

# Check bvals and bvecs shapes
check_bvals_shape(bvals_file, num_volumes)
check_bvecs_shape(bvecs_file, num_volumes)
```
