```python
import nibabel as nib
import numpy as np
import os
from nipype.interfaces.fsl import FLIRT, Info
import logging
from scipy.fft import fftn, ifftn, fftshift

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Paths to the files (adjust these paths)
# asd_anat_path = 'path/to/asd/anat.nii.gz'
# asd_rest_path = 'path/to/asd/rest.nii.gz'
# adhd_anat_path = 'path/to/adhd/anat.nii.gz'
# adhd_rest_path = 'path/to/adhd/rest.nii.gz'

# Output directory
output_dir = '/Users/swapnilbhattacharya/Desktop/Dissertation_Code/Output'
os.makedirs(output_dir, exist_ok=True)

def load_image(file_path):
    """ Load an MRI image and return the image object """
    try:
        img = nib.load(file_path)
        logger.info(f"Loaded image from {file_path} with shape {img.shape}")
        return img
    except Exception as e:
        logger.error(f"Error loading image from {file_path}: {e}")
        return None

def save_image(img, file_path):
    """ Save an MRI image to a specified path """
    try:
        nib.save(img, file_path)
        logger.info(f"Saved image to {file_path}")
    except Exception as e:
        logger.error(f"Error saving image to {file_path}: {e}")

def normalize_and_register(input_img, reference_img, output_path):
    """ Normalize and register an image to a standard template """
    try:
        flirt = FLIRT(bins=640, cost_func='mutualinfo')
        flirt.inputs.in_file = input_img
        flirt.inputs.reference = reference_img
        flirt.inputs.out_file = output_path
        flirt.inputs.out_matrix_file = output_path.replace('.nii.gz', '.mat')
        result = flirt.run()
        logger.info(f"Normalized and registered image saved to {output_path}")
        return result.outputs.out_file
    except Exception as e:
        logger.error(f"Error during normalization and registration: {e}")
        return None

def fft_smooth_image(input_img, output_path, sigma=1):
    """ Apply FFT-based smoothing to an image """
    try:
        img = nib.load(input_img)
        data = img.get_fdata()
        
        # Perform FFT
        fft_data = fftn(data)
        
        # Create Gaussian filter in frequency domain
        dims = data.shape
        freq = [np.fft.fftfreq(d) for d in dims]
        grid = np.meshgrid(*freq, indexing='ij')
        gauss_filter = np.exp(-0.5 * sum([(g/sigma)**2 for g in grid]))
        
        # Apply filter
        fft_data *= fftshift(gauss_filter)
        
        # Inverse FFT to get smoothed image
        smooth_data = np.real(ifftn(fft_data))
        
        # Save the smoothed image
        smooth_img = nib.Nifti1Image(smooth_data, img.affine)
        save_image(smooth_img, output_path)
        
        logger.info(f"FFT-smoothed image saved to {output_path}")
        return output_path
    except Exception as e:
        logger.error(f"Error during FFT smoothing: {e}")
        return None

def preprocess_mri(anat_path, rest_path, output_dir, subject_type, smoothing=False, sigma=1):
    """ Preprocess MRI data: normalize, register, and optionally smooth """
    # Load images
    anat_img = load_image(anat_path)
    rest_img = load_image(rest_path)
    
    if anat_img is None or rest_img is None:
        logger.error("Failed to load one or more images. Exiting.")
        return
    
    # Define output paths
    anat_output_path = os.path.join(output_dir, f'{subject_type}_anat_normalized.nii.gz')
    rest_output_path = os.path.join(output_dir, f'{subject_type}_rest_normalized.nii.gz')
    
    # Normalization and registration
    mni_template = Info.standard_image('MNI152_T1_1mm_brain.nii.gz')
    anat_normalized = normalize_and_register(anat_path, mni_template, anat_output_path)
    rest_normalized = normalize_and_register(rest_path, anat_output_path, rest_output_path)  # Register to the anatomical
    
    if smoothing:
        # Apply FFT-based smoothing
        anat_smoothed_path = os.path.join(output_dir, f'{subject_type}_anat_smoothed.nii.gz')
        rest_smoothed_path = os.path.join(output_dir, f'{subject_type}_rest_smoothed.nii.gz')
        
        anat_smoothed = fft_smooth_image(anat_normalized, anat_smoothed_path, sigma=sigma)
        rest_smoothed = fft_smooth_image(rest_normalized, rest_smoothed_path, sigma=sigma)
        
        return anat_smoothed, rest_smoothed
    
    return anat_normalized, rest_normalized

# Preprocess ASD and ADHD datasets
asd_anat_normalized, asd_rest_normalized = preprocess_mri(asd_anat_path, asd_rest_path, output_dir, 'ASD', smoothing=True, sigma=1)
adhd_anat_normalized, adhd_rest_normalized = preprocess_mri(adhd_anat_path, adhd_rest_path, output_dir, 'ADHD', smoothing=True, sigma=1)
```

```python
import nibabel as nib
import numpy as np
import pandas as pd
import os
import logging
from nilearn import image, plotting
from dipy.io import read_bvals_bvecs
from dipy.core.gradients import gradient_table
from dipy.reconst.dti import TensorModel
from dipy.reconst.dti import fractional_anisotropy, color_fa

# Configure logging for debugging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Function to check and correct bvals and bvecs
def check_and_correct_bvals_bvecs(dti_path, bval_path, bvec_path):
    # Load DTI data
    dti_img = nib.load(dti_path)
    dti_data = dti_img.get_fdata()
    
    # Load b-values and b-vectors
    bvals, bvecs = read_bvals_bvecs(bval_path, bvec_path)
    
    # Check for shape mismatch
    if dti_data.shape[-1] != bvals.shape[0] or dti_data.shape[-1] != bvecs.shape[0]:
        logger.error(f"Mismatch between number of DTI volumes ({dti_data.shape[-1]}) and bvals/bvecs (bvals: {bvals.shape[0]}, bvecs: {bvecs.shape[0]})")
        # Assuming one volume is missing in bvals and bvecs
        missing_bval = 0  # Example value, replace with actual if known
        missing_bvec = [0, 0, 0]  # Example vector, replace with actual if known
        
        bvals = np.append(bvals, missing_bval)
        bvecs = np.vstack([bvecs, missing_bvec])
        logger.info("Corrected the shape mismatch by adding missing b-value and b-vector")

    return bvals, bvecs, dti_data, dti_img

# Check and correct bvals and bvecs for ASD DTI
bvals, bvecs, asd_dti_data, asd_dti_img = check_and_correct_bvals_bvecs(asd_dti_path, asd_bval_path, asd_bvec_path)

# Load and preprocess ASD anatomical MRI
logger.info(f"Loading ASD anatomical MRI from {asd_anat_path}")
asd_anat_img = nib.load(asd_anat_path)
asd_anat_data = asd_anat_img.get_fdata()
logger.info(f"ASD anatomical MRI shape: {asd_anat_data.shape}")

# Load and preprocess ASD resting-state fMRI
logger.info(f"Loading ASD resting-state fMRI from {asd_rest_path}")
asd_rest_img = nib.load(asd_rest_path)
asd_rest_data = asd_rest_img.get_fdata()
logger.info(f"ASD resting-state fMRI shape: {asd_rest_data.shape}")

# Fit the DTI model for ASD data
logger.info("Fitting the DTI model for ASD data")
gtab = gradient_table(bvals, bvecs)
tensor_model = TensorModel(gtab)
tensor_fit = tensor_model.fit(asd_dti_data)

# Extract Fractional Anisotropy (FA) and color FA for ASD data
logger.info("Extracting FA and color FA for ASD data")
fa = fractional_anisotropy(tensor_fit.evals)
fa_img = nib.Nifti1Image(fa, asd_dti_img.affine)
color_fa_img = color_fa(fa, tensor_fit.evecs)
color_fa_img = nib.Nifti1Image(color_fa_img, asd_dti_img.affine)

# Load and preprocess ADHD anatomical MRI
logger.info(f"Loading ADHD anatomical MRI from {adhd_anat_path}")
adhd_anat_img = nib.load(adhd_anat_path)
adhd_anat_data = adhd_anat_img.get_fdata()
logger.info(f"ADHD anatomical MRI shape: {adhd_anat_data.shape}")

# Load and preprocess ADHD resting-state fMRI
logger.info(f"Loading ADHD resting-state fMRI from {adhd_rest_path}")
adhd_rest_img = nib.load(adhd_rest_path)
adhd_rest_data = adhd_rest_img.get_fdata()
logger.info(f"ADHD resting-state fMRI shape: {adhd_rest_data.shape}")

# Example feature extraction function (to be customized as needed)
def extract_features(anat_data, rest_data):
    logger.info("Extracting features from anatomical and resting-state MRI data")
    # Placeholder for feature extraction logic
    features = {
        'mean_anat': np.mean(anat_data),
        'std_anat': np.std(anat_data),
        'mean_rest': np.mean(rest_data),
        'std_rest': np.std(rest_data),
    }
    return features

# Extract features for ASD and ADHD data
logger.info("Extracting features for ASD data")
asd_features = extract_features(asd_anat_data, asd_rest_data)

logger.info("Extracting features for ADHD data")
adhd_features = extract_features(adhd_anat_data, adhd_rest_data)

# Create dataframes
logger.info("Creating dataframes for ASD and ADHD features")
asd_df = pd.DataFrame([asd_features])
asd_df['group'] = 'ASD'

adhd_df = pd.DataFrame([adhd_features])
adhd_df['group'] = 'ADHD'

# Combine dataframes
combined_df = pd.concat([asd_df, adhd_df], ignore_index=True)

# Save the combined dataframe to a CSV file
output_path = 'combined_mri_features.csv'
logger.info(f"Saving combined dataframe to {output_path}")
combined_df.to_csv(output_path, index=False)

logger.info("Dataset creation complete")
```
